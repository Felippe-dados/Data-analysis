```python
import pandas as pd

df = pd.read_csv('/base_vendas_10000_variada.csv')

df["Data"] = pd.to_datetime(df["Data"])

display(df)
```

```python
import plotly.express as plt

fa = df.groupby('Produto')['Valor Total'].sum()
fa = fa.reset_index() # Convert the Series to a DataFrame

bar = plt.bar(fa,x = 'Produto',y = 'Valor Total')
bar.show()
```

```python
import plotly.express as plt

f = df.groupby('Produto')['Quantidade'].sum().reset_index()

bar = plt.bar(f,x = 'Produto',y = 'Quantidade')
bar.show()
```

```python
import plotly.express as plt

faturamento_por_data = df.groupby('Data')['Valor Total'].sum()

faturamento_mensal = faturamento_por_data.resample("ME").sum().reset_index()

bar = plt.bar(faturamento_mensal, x = 'Data', y = 'Valor Total')
bar.show()
```

```python
import plotly.express as plt

faturamento_por_regiao = df.groupby('Região')['Valor Total'].sum()

faturamento_por_regiao = faturamento_por_regiao.reset_index()

bar = plt.bar(faturamento_por_regiao, x = 'Região', y = 'Valor Total')
bar.show()
```
